```python
import numpy as np
import abaudit as ab

rng  = np.random.default_rng(42)
ctrl = rng.normal(0.0, 1.0, 500)
trt  = rng.normal(0.3, 1.0, 500)

result = ab.audit(
    ctrl, trt,
    prior_f  = 0.2,
    metrics  = ['conversion', 'revenue', 'session_time'],
    n_peeks  = 5,
)
result.summary()
```
