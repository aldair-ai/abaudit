```python
import numpy as np
import abaudit as ab

rng  = np.random.default_rng(42)
ctrl = rng.normal(0, 1, 500)
trt  = rng.normal(0.3, 1, 500)

# During-experiment checks
ab.check_srm(n_control=490, n_treatment=510)
ab.check_optional_stopping([0.12, 0.08, 0.04, 0.06, 0.03])

# Full audit + HTML report
result = ab.audit(ctrl, trt, prior_f=0.2,
                  metrics=["conversion", "revenue"],
                  n_peeks=5)
result.summary()
ab.generate_report(result, "audit_report.html")
# open audit_report.html in your browser
```
